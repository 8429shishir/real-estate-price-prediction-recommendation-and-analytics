```python
def change_in_cr(x):
    # If x is numeric, return as float
    if isinstance(x, (float, int)):
        return x

    # Handle lists (after split)
    if isinstance(x, list):
        if len(x) == 1:  # e.g. ["nan"] or ["2500000.0"]
            try:
                return float(x[0])
            except:
                return None
        elif len(x) >= 2:
            num_str, unit = x[0], x[1]
            try:
                num = float(num_str)
            except:
                return None

            if unit == 'Lac':
                return round(num / 100, 2)
            else:
                return round(num, 2)
        else:
            return None

    # Handle strings (if .str.split not used)
    if isinstance(x, str):
        parts = x.split()
        return change_in_cr(parts)

    return None

       
            
```

```python
df['price']=df['price'].astype(str).str.split(' ').apply(change_in_cr)
```
